````python
# Run this first. It's the test helper used in every test cell.
def check(fn, *args, expected):
    label = f"{fn.__name__}{args if len(args) != 1 else '(' + repr(args[0]) + ')'}"
    try:
        got = fn(*args)
    except Exception as e:
        print(f"FAIL  {label} crashed: {type(e).__name__}: {e}")
        return
    if isinstance(expected, float) and isinstance(got, (int, float)):
        ok = abs(got - expected) < 1e-9
    else:
        ok = got == expected
    if ok:
        print(f"PASS  {label} == {expected!r}")
    else:
        print(f"FAIL  {label} should be {expected!r}, but got {got!r}")

print("Helper loaded.")
````

````python
# Solutions (run to check they pass)
def square(x: int) -> int:
    return x * x

def area_rectangle(width: float, height: float) -> float:
    return width * height

def price_with_discount(price: float, percent: float) -> float:
    return price * (1 - percent / 100)

def average(numbers: list[float]) -> float:
    total = 0
    for n in numbers:
        total += n
    return total / len(numbers)

def grade_label(score: float) -> str:
    if score >= 5.5:
        return "pass"
    else:
        return "fail"

def sum_list(numbers: list[int]) -> int:
    total = 0
    for n in numbers:
        total += n
    return total          # outside the loop

# E2
total = make_total(4, 5)
print(total)

def count_even(numbers: list[int]) -> int:
    count = 0
    for n in numbers:
        if n % 2 == 0:
            count += 1
    return count

def largest(numbers: list[int]) -> int:
    biggest = numbers[0]
    for n in numbers:
        if n > biggest:
            biggest = n
    return biggest

def factorial(n: int) -> int:
    result = 1
    for i in range(1, n + 1):
        result = result * i
    return result

def to_fahrenheit(temps: list[float]) -> list[float]:
    result = []
    for c in temps:
        result.append(c * 9 / 5 + 32)
    return result

def count_passed(scores: list[float]) -> int:
    count = 0
    for s in scores:
        if grade_label(s) == "pass":
            count += 1
    return count

def shopping_total(prices: list[float], percent: float) -> float:
    total = 0
    for p in prices:
        total += price_with_discount(p, percent)
    return round(total, 2)
````
