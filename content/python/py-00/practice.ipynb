````python
# Run this cell first. It defines check(), the test helper used in every test cell.
def check(fn, *args, expected, label=None):
    """Call fn(*args), compare with expected and print PASS or FAIL. It never crashes."""
    if label is None:
        shown = ", ".join(repr(a) for a in args)
        label = f"{getattr(fn, '__name__', 'call')}({shown})"
    try:
        got = fn(*args)
    except Exception as error:
        print(f"FAIL  {label} crashed: {type(error).__name__}: {error}")
        return
    if isinstance(expected, float) and isinstance(got, (int, float)) and not isinstance(got, bool):
        ok = abs(got - expected) < 1e-9
    else:
        # True is not 1 here: a function that should return a bool has to return a bool
        ok = got == expected and isinstance(got, bool) == isinstance(expected, bool)
    if ok:
        print(f"PASS  {label} == {expected!r}")
    else:
        print(f"FAIL  {label} should be {expected!r}, but got {got!r}")

print("Helper loaded.")
````

### Exercise 6 · Write from scratch: from loop program to function

The recipe for turning a loop program into a function:
1. Put the loop inside `def name(...):`.
2. The data it works on becomes a **parameter**.
3. The final `print(...)` becomes `return ...`.
4. Call it and **store** the result.

Turn this loop program into `count_even(numbers)`:

```python
numbers = [3, 8, 10, 7, 2]
count = 0
for n in numbers:
    if n % 2 == 0:
        count += 1
print(count)
```

````python
def count_even(numbers: list[int]) -> int:
    """Return how many of the numbers are even."""
    pass
````

````python
check(count_even, [3, 8, 10, 7, 2], expected=3)
check(count_even, [1, 3, 5], expected=0)
check(count_even, [], expected=0)
````

<details><summary>Hint 1</summary>

The list is already the parameter, so the line `numbers = [...]` is not needed inside the function.
</details>

<details><summary>Hint 2</summary>

Keep `count = 0` and the loop as they are, indented inside the function. Only the last line changes.
</details>

<details><summary>Show solution</summary>

```python
def count_even(numbers: list[int]) -> int:
    """Return how many of the numbers are even."""
    count = 0
    for n in numbers:
        if n % 2 == 0:
            count += 1
    return count
```
</details>

<details><summary>Hint 1</summary>

```python
def largest(numbers: list[int]) -> int:
    """Return the biggest number in a list that is not empty."""
    biggest = numbers[0]
    for n in numbers:
        if n > biggest:
            biggest = n
    return biggest
```

    return result
